```python
from pathlib import Path
from rosbags.highlevel import AnyReader
from rosbags.serde import deserialize_cdr
from rosbags.typesys import Stores, get_types_from_msg, get_typestore

# Define custom message types

LOWCMD_MSG = """\
uint8[2] head
uint8 level_flag
uint8 frame_reserve
uint32[2] sn
uint32[2] version
uint16 bandwidth
MotorCmd[20] motor_cmd
BmsCmd bms_cmd
uint8[40] wireless_remote
uint8[12] led
uint8[2] fan
uint8 gpio
uint32 reserve
uint32 crc
"""

MOTORCMD_MSG = """\
uint8 mode
float32 q
float32 dq
float32 tau
float32 kp
float32 kd
uint32[3] reserve
"""

BMSCMD_MSG = """\
uint8 off
uint8[3] reserve
"""

LOWSTATE_MSG = """\
uint8[2] head
uint8 level_flag
uint8 frame_reserve
uint32[2] sn
uint32[2] version
uint16 bandwidth
unitree_go/msg/IMUState imu_state
unitree_go/msg/MotorState[20] motor_state
unitree_go/msg/BmsState bms_state
int16[4] foot_force
int16[4] foot_force_est
uint32 tick
uint8[40] wireless_remote
uint8 bit_flag
float32 adc_reel
int8 temperature_ntc1
int8 temperature_ntc2
float32 power_v
float32 power_a
uint16[4] fan_frequency 
uint32 reserve
uint32 crc
"""

IMUSTATE_MSG = """\
float32[4] quaternion
float32[3] gyroscope
float32[3] accelerometer
float32[3] rpy
int8 temperature
"""

MOTORSTATE_MSG = """\
uint8 mode
float32 q
float32 dq
float32 ddq
float32 tau_est
float32 q_raw
float32 dq_raw
float32 ddq_raw
int8 temperature
uint32 lost
uint32[2] reserve
"""

BMSSTATE_MSG = """\
uint8 version_high
uint8 version_low
uint8 status
uint8 soc
int32 current
uint16 cycle
int8[2] bq_ntc
int8[2] mcu_ntc
uint16[15] cell_vol
"""

# Initialize Typestore and Register Messages
typestore = get_typestore(Stores.ROS2_FOXY)
typestore.register(get_types_from_msg(MOTORCMD_MSG, 'unitree_go/msg/MotorCmd'))
typestore.register(get_types_from_msg(BMSCMD_MSG, 'unitree_go/msg/BmsCmd'))
typestore.register(get_types_from_msg(LOWCMD_MSG, 'unitree_go/msg/LowCmd'))

typestore.register(get_types_from_msg(LOWSTATE_MSG, 'unitree_go/msg/LowState'))
typestore.register(get_types_from_msg(IMUSTATE_MSG, 'unitree_go/msg/IMUState'))
typestore.register(get_types_from_msg(MOTORSTATE_MSG, 'unitree_go/msg/MotorState'))
typestore.register(get_types_from_msg(BMSSTATE_MSG, 'unitree_go/msg/BmsState'))
assert 'unitree_go/msg/LowState' in typestore.types, "❌ LowState message type not registered!"
# Assign Types
LowState = typestore.types['unitree_go/msg/LowState']
LowCmd = typestore.types['unitree_go/msg/LowCmd']

# Path to the ROS2 bag file
bag_path = Path("bags/walking_1_0311")  # Update with the actual bag path
data_list = []
cmd_list = []
data_time_list = []
cmd_time_list = []
count = 0 
st = 0
# Read the ROS2 bag and deserialize messages
with AnyReader([bag_path]) as reader:
    print("\nAvailable topics and message types in the bag:")
    for conn in reader.connections:
        print(f"Topic: {conn.topic}, Type: {conn.msgtype}")

    topic_to_read = "/lowstate"  # Replace with actual topic name

    topic_to_read_cmd = "/lowcmd"

    # print(f"\nReading messages from {topic_to_read}...\n")
    for connection, timestamp, rawdata in reader.messages():
        if connection.topic == topic_to_read:
            msg = deserialize_cdr(rawdata, 'unitree_go/msg/LowState', typestore)
            data_list.append(msg)
            data_time_list.append(timestamp)
    
    for connection, timestamp, rawdata in reader.messages():
        if connection.topic == topic_to_read_cmd:
            msg = deserialize_cdr(rawdata, 'unitree_go/msg/LowCmd', typestore)
            cmd_list.append(msg)
            cmd_time_list.append(timestamp)
```
